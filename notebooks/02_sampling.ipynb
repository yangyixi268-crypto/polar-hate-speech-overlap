```python
import pandas as pd

TYPES = ['political', 'racial/ethnic', 'religious', 'gender/sexual', 'other']
MANIF = ['stereotype', 'vilification', 'dehumanization',
         'extreme_language', 'lack_of_empathy', 'invalidation']
SEED = 42

# 1. Load the English test split
test = pd.read_csv('../data-public/test/eng.csv').drop(columns=['canary'])

# 2. Sample 150 polarized and 150 non-polarized instances
pol = test[test['polarization'] == 1].sample(150, random_state=SEED)
non = test[test['polarization'] == 0].sample(150, random_state=SEED)

# 3. Combine and shuffle, so annotators do not see all polarized instances in a row
sample = pd.concat([pol, non]).sample(frac=1, random_state=SEED).reset_index(drop=True)

# 4. Save
sample.to_csv('../data/sample_300.csv', index=False)          # full data, NOT uploaded to GitHub
sample[['id']].to_csv('../data/sample_ids.csv', index=False)  # IDs only, safe to upload

# 5. Check
print(f'{len(sample)} instances, {sample["polarization"].sum()} polarized')
print(sample[TYPES + MANIF].sum().to_string())
```
